# Node Class

```python
import numpy as np

class Node():
    def __init__(self, feature_index=None, thresholds=None, children=None, info_gain=None, value=None):
        self.feature_index = feature_index if feature_index is not None else -1
        self.thresholds = thresholds if thresholds is not None else []
        self.children = children if children is not None else []
        self.info_gain = info_gain if info_gain is not None else 0.0
        self.value = value

class DecisionTreeClassifier():
    def __init__(self, min_samples_split=2, max_depth=2, num_splits=3, criterion="entropy"):
        self.root = None
        self.min_samples_split = max(2, min_samples_split)
        self.max_depth = max(1, max_depth)
        self.num_splits = max(2, num_splits)
        self.criterion = criterion.lower()
    def build_tree(self, dataset, curr_depth=0):
      X, Y = dataset[:, :-1], dataset[:, -1]
      num_samples, num_features = np.shape(X)

      if num_samples < self.min_samples_split or curr_depth >= self.max_depth:
          leaf_value = self.calculate_leaf_value(Y)
          return Node(value=leaf_value)

      if self.criterion == "entropy":
          best_split = self.get_best_split_info_gain(dataset, num_samples, num_features)
      else:
          best_split = self.get_best_split_gini_index(dataset, num_samples, num_features)


      if best_split["datasets"] is not None and best_split["info_gain"] > 0.0001:
          children = []
          for subset in best_split["datasets"]:
              if len(subset) > 0:
                  child = self.build_tree(subset, curr_depth + 1)
                  children.append(child)

          if len(children) > 1:
              return Node(feature_index=best_split["feature_index"],
                        thresholds=best_split["thresholds"],
                        children=children,
                        info_gain=best_split["info_gain"])


      leaf_value = self.calculate_leaf_value(Y)
      return Node(value=leaf_value)

    def get_best_split_info_gain(self, dataset, num_samples, num_features):
        best_split = {"info_gain": -float("inf"), "feature_index": None, "thresholds": None, "datasets": None}

        for feature_index in range(num_features):
            feature_values = dataset[:, feature_index]
            unique_values = np.unique(feature_values)

            if len(unique_values) < 2:
                continue

            possible_thresholds = np.percentile(feature_values, np.linspace(0, 100, self.num_splits + 1)[1:-1])

            for threshold in possible_thresholds:
                datasets = self.split(dataset, feature_index, [threshold])
                if len(datasets) >= 2:
                    y_parent = dataset[:, -1]
                    y_children = [subset[:, -1] for subset in datasets]

                    if all(len(np.unique(y)) <= 1 for y in y_children):
                        continue

                    info_gain = self.information_gain(y_parent, y_children, mode="entropy")

                    if info_gain > best_split["info_gain"]:
                        best_split["feature_index"] = feature_index
                        best_split["thresholds"] = [threshold]
                        best_split["datasets"] = datasets
                        best_split["info_gain"] = info_gain

        return best_split

    def get_best_split_gini_index(self, dataset, num_samples, num_features):
        best_split = {"info_gain": float("inf"), "feature_index": None, "thresholds": None, "datasets": None}

        for feature_index in range(num_features):
            feature_values = dataset[:, feature_index]
            unique_values = np.unique(feature_values)

            if len(unique_values) < 2:
                continue

            possible_thresholds = np.percentile(feature_values, np.linspace(0, 100, self.num_splits + 1)[1:-1])

            for threshold in possible_thresholds:
                datasets = self.split(dataset, feature_index, [threshold])
                if len(datasets) >= 2:
                    y_parent = dataset[:, -1]
                    y_children = [subset[:, -1] for subset in datasets]

                    if all(len(np.unique(y)) <= 1 for y in y_children):
                        continue

                    gini = self.gini_index_split(y_parent, y_children)

                    if gini < best_split["info_gain"]:
                        best_split["feature_index"] = feature_index
                        best_split["thresholds"] = [threshold]
                        best_split["datasets"] = datasets
                        best_split["info_gain"] = gini

        return best_split

    def split(self, dataset, feature_index, thresholds):
        subsets = []
        feature_values = dataset[:, feature_index]

        if len(thresholds) == 0:
            return [dataset]

        for i in range(len(thresholds)):
            if i == 0:
                indices = feature_values <= thresholds[i]
            else:
                indices = (feature_values > thresholds[i-1]) & (feature_values <= thresholds[i])

            subset = dataset[indices]
            if len(subset) > 0:
                subsets.append(subset)

        indices = feature_values > thresholds[-1]
        subset = dataset[indices]
        if len(subset) > 0:
            subsets.append(subset)

        return subsets

    def information_gain(self, parent, children, mode="entropy"):
        if len(parent) == 0:
            return 0

        parent_size = len(parent)
        entropy_children = sum([(len(child) / parent_size) * self.entropy(child) for child in children if len(child) > 0])

        if mode == "entropy":
            gain = self.entropy(parent) - entropy_children
        else:
            gain = self.gini_index(parent) - entropy_children

        return gain

    def gini_index_split(self, parent, children):
        parent_size = len(parent)
        gini_children = sum([(len(child) / parent_size) * self.gini_index(child) for child in children if len(child) > 0])
        return gini_children

    def entropy(self, y):
        if len(y) == 0:
            return 0

        class_labels = np.unique(y)
        entropy = 0
        for cls in class_labels:
            p_cls = len(y[y == cls]) / len(y)
            if p_cls > 0:
                entropy += -p_cls * np.log2(p_cls)
        return entropy

    def gini_index(self, y):
        if len(y) == 0:
            return 0

        class_labels = np.unique(y)
        gini = 1
        for cls in class_labels:
            p_cls = len(y[y == cls]) / len(y)
            gini -= p_cls ** 2
        return gini

    def calculate_leaf_value(self, Y):
        if len(Y) == 0:
            return 0
        unique_classes, counts = np.unique(Y, return_counts=True)
        return unique_classes[np.argmax(counts)]

    def fit(self, X, Y):
        X = np.array(X)
        Y = np.array(Y).reshape(-1)
        dataset = np.column_stack((X, Y))
        self.root = self.build_tree(dataset)

    def predict(self, X):
        X = np.array(X)
        return np.array([self.make_prediction(x, self.root) for x in X])

    def make_prediction(self, x, tree):
        if tree.value is not None:
            return tree.value

        if tree.feature_index < 0 or tree.feature_index >= len(x):
            return 0

        feature_val = x[tree.feature_index]

        for i, threshold in enumerate(tree.thresholds):
            if feature_val <= threshold:
                if i < len(tree.children):
                    return self.make_prediction(x, tree.children[i])
                return 0

        if tree.children:
            return self.make_prediction(x, tree.children[-1])
        return 0

def accuracy_score(y_true, y_pred):
    y_true = np.array(y_true).reshape(-1)
    y_pred = np.array(y_pred).reshape(-1)

    if len(y_true) != len(y_pred):
        raise ValueError("Length of y_true and y_pred must match")

    return np.mean(y_true == y_pred)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

classifier = DecisionTreeClassifier(min_samples_split=150, max_depth=17, num_splits=3,criterion='gini')
classifier.fit(X_train, y_train)

train_predictions = classifier.predict(X_train)
test_predictions = classifier.predict(X_test)

train_accuracy = accuracy_score(y_train, train_predictions)
test_accuracy = accuracy_score(y_test, test_predictions)

print('Train Accuracy:', train_accuracy)
print('Test Accuracy:', test_accuracy)
```
